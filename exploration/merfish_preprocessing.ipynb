## Small dataset

```python
import numpy as np
import scanpy as sc
```

```python
sequential_genes = [
    "Oxt", "Penk", "Sst", "Tac1", "Gal", "Cartpt", "Vgf", "Trh", 
    "Nts", "Scg2", "Gnrh1", "Tac2", "Cck", "Crh", "Ucn3", 
    "Adcyap1", "Nnat", "Sln", "Mbp", "Th", "Fos"
]
overlap_starmap = ['Sst', 'Cck', 'Fos', 'Mbp']
```

```python
import numpy as np
import scanpy as sc

for i in range(1, 6):
    adata = sc.read_h5ad(f"./data/merfish/merfish_slice_{i}.h5ad")
    
    removed_adata = adata[:, ~adata.var_names.isin(sequential_genes)].copy()
    sc.pp.normalize_total(removed_adata, target_sum=1e4, inplace=True)

    removed_adata.X = np.rint(removed_adata.X).astype(np.int32)
    removed_adata.write_h5ad(f"./data/merfish/merfish_slice_{i}_int.h5ad")

```

```python
adata1 = sc.read_h5ad("./data/merfish/merfish_slice_1_int.h5ad")
adata2 = sc.read_h5ad("./data/merfish/merfish_slice_2_int.h5ad")
adata3 = sc.read_h5ad("./data/merfish/merfish_slice_3_int.h5ad")
adata4 = sc.read_h5ad("./data/merfish/merfish_slice_4_int.h5ad")
adata5 = sc.read_h5ad("./data/merfish/merfish_slice_5_int.h5ad")

adata_all = adata1.concatenate(adata2, adata3, adata4, adata5)
adata_all.write_h5ad("./data/merfish/merfish_all_int.h5ad")
```
